```python
catalog = dbutils.widgets.get("catalog")
schema_prefix = dbutils.widgets.get("schema_prefix")
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"USE SCHEMA {schema_prefix}_gold")
```

```python
from pyspark.sql.functions import count, sum, avg

(spark.read
  .table(f"{catalog}.{schema_prefix}_silver.crm_silver")
  .groupBy("stage")
  .agg(
    count("lead_id").alias("lead_count"),
    sum("opportunity_value").alias("total_opportunity_value"),
    avg("opportunity_value").alias("avg_opportunity_value")
  )
  .write
  .mode("overwrite")
  .saveAsTable("crm_gold")
)
```
