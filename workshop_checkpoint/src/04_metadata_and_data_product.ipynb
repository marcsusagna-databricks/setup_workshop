```python
catalog = dbutils.widgets.get("catalog")
schema_prefix = dbutils.widgets.get("schema_prefix")
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"USE SCHEMA {schema_prefix}_silver")
```

## Tagging and metadata

```python
table_name = f"{catalog}.{schema_prefix}_silver.crm_silver"

# Add table description
spark.sql(f"COMMENT ON TABLE {table_name} IS 'Silver CRM table containing cleaned and enriched lead data from the CRM system'")

# Add column comments on the first 3 columns
spark.sql(f"ALTER TABLE {table_name} ALTER COLUMN lead_id COMMENT 'Unique identifier for each lead'")
spark.sql(f"ALTER TABLE {table_name} ALTER COLUMN company COMMENT 'Company to which the lead is associated'")
spark.sql(f"ALTER TABLE {table_name} ALTER COLUMN contact_name COMMENT 'Full name of the contact person for the lead'")

# Add PII tag on contact_name column
spark.sql(f"ALTER TABLE {table_name} ALTER COLUMN contact_name SET TAGS ('demo_pii_data' = 'true')")

print(f"Metadata added successfully to {table_name}")
```

## PK and FK

```python
table_name = f"{catalog}.{schema_prefix}_silver.crm_silver"

# Set lead_id as NOT NULL (required for primary key)
spark.sql(f"ALTER TABLE {table_name} ALTER COLUMN lead_id SET NOT NULL")

# Add primary key constraint on lead_id
spark.sql(f"ALTER TABLE {table_name} ADD CONSTRAINT crm_silver_pk PRIMARY KEY (lead_id)")

print(f"Primary key constraint added on lead_id for {table_name}")
```
